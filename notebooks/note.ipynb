# CHIẾN LƯỢC PHÂN TÍCH DỮ LIỆU HỘ NGHÈO PHỤC VỤ MỤC TIÊU GIẢM NGHÈO BỀN VỮNG TẠI TỈNH ĐẮK NÔNG

## 1. Phân tích Mô tả – Descriptive Analysis

### Mục tiêu

Phân tích mô tả được sử dụng để khái quát quy mô dữ liệu, cơ cấu hộ nghèo – hộ cận nghèo và đặc điểm độ tuổi của các nhân khẩu trong tập dữ liệu.

Tầng này trả lời các câu hỏi:

- Tập dữ liệu có bao nhiêu bản ghi hợp lệ?
- Hộ nghèo và hộ cận nghèo phân bố như thế nào?
- Nhân khẩu trong hai nhóm hộ có độ tuổi trung bình và trung vị bao nhiêu?
- Nhóm trẻ em, lao động và người lớn tuổi chiếm tỷ lệ như thế nào?
- Các nhóm hộ tập trung tại huyện và dân tộc nào?

### 1.1. Mô tả cơ cấu phân loại hộ

Tần suất và tỷ lệ của hai nhãn `Hộ nghèo` và `Hộ cận nghèo` được thống kê trên toàn bộ dữ liệu sau lọc.

Biểu đồ cột hoặc biểu đồ Donut được sử dụng để thể hiện cơ cấu hai nhóm. Kết quả này cung cấp cái nhìn tổng quan về mức độ cân bằng của biến mục tiêu trước khi thực hiện các phân tích sâu hơn.

### 1.2. Mô tả dữ liệu năm sinh và tuổi

Từ biến `NGAY_SINH`, nghiên cứu thống kê:

- Số quan sát hợp lệ.
- Số giá trị thiếu.
- Tuổi trung bình.
- Tuổi trung vị.
- Độ lệch chuẩn.
- Tuổi nhỏ nhất và lớn nhất.
- Các phân vị của phân phối tuổi.

Kết quả cho thấy:

- Nhóm hộ nghèo có độ tuổi trung bình khoảng **28,3 tuổi**, trung vị **23 tuổi**.
- Nhóm hộ cận nghèo có độ tuổi trung bình khoảng **30,2 tuổi**, trung vị **27 tuổi**.
- Tỷ lệ người dưới 18 tuổi trong nhóm hộ nghèo cao hơn nhóm hộ cận nghèo.
- Tỷ lệ người từ 50 tuổi trở lên trong nhóm hộ nghèo không cao hơn nhóm hộ cận nghèo trên toàn bộ dữ liệu nhân khẩu.

Như vậy, dữ liệu mô tả ban đầu không ủng hộ giả thuyết đơn giản rằng hộ nghèo chủ yếu có cơ cấu dân số già hơn. Ngược lại, nhóm hộ nghèo có xu hướng tập trung nhiều hơn ở các nhóm tuổi trẻ.

### 1.3. Mô tả đặc điểm chủ hộ

Các bản ghi có `QUANHE_CHUHO == "Chủ hộ"` được tách riêng để phân tích:

- Tuổi trung bình và trung vị của chủ hộ.
- Cơ cấu giới tính chủ hộ.
- Nhóm tuổi của chủ hộ.
- Sự khác biệt về tuổi chủ hộ giữa các huyện.
- Sự khác biệt giữa hộ nghèo và hộ cận nghèo.

Biểu đồ hộp được sử dụng để thể hiện phân phối tuổi chủ hộ theo huyện và phân loại hộ.

### 1.4. Mô tả theo địa lý và dân tộc

Dữ liệu được tổng hợp theo:

- `HUYEN`
- `XA`
- `DAN_TOC`
- `PHAN_LOAI_HO`

Mục tiêu là xác định những địa bàn và nhóm dân tộc có số lượng hoặc tỷ lệ hộ nghèo cao.

Kết quả mô tả cho thấy mức độ nghèo không phân bố đồng đều giữa các nhóm dân tộc. Trong các nhóm được khảo sát, dân tộc Dao có tỷ lệ thuộc hộ nghèo cao hơn đáng kể so với nhiều nhóm còn lại. Tuy nhiên, đây mới là khác biệt quan sát được và chưa thể được diễn giải như quan hệ nhân quả.

## 2. Phân tích Khám phá – Exploratory Data Analysis

### Mục tiêu

Phân tích khám phá được sử dụng để nhận diện hình dạng phân phối, các nhóm có sự khác biệt đáng chú ý và những mối liên hệ tiềm năng giữa cơ cấu tuổi với trạng thái hộ nghèo.

Tầng này không nhằm chứng minh nguyên nhân mà tập trung phát hiện các mẫu dữ liệu cần được kiểm định ở tầng chẩn đoán.

### 2.1. Khám phá phân phối tuổi theo phân loại hộ

Phân phối tuổi của hai nhóm hộ được so sánh bằng:

- Histogram.
- Biểu đồ mật độ KDE.
- Biểu đồ hộp.
- Các chỉ số trung bình, trung vị và phân vị.

Kết quả cho thấy đường phân phối của nhóm hộ nghèo dịch về các nhóm tuổi trẻ hơn so với hộ cận nghèo. Nhóm hộ nghèo có tỷ lệ trẻ em dưới 18 tuổi cao hơn, trong khi hộ cận nghèo có tuổi trung bình và trung vị cao hơn.

Dữ liệu vì vậy gợi ý rằng **gánh nặng nhân khẩu học ở nhóm hộ nghèo có thể liên quan nhiều hơn đến tỷ lệ trẻ em và người phụ thuộc**, thay vì chỉ liên quan đến già hóa dân số.

### 2.2. Khám phá theo nhóm tuổi 5 năm

Tuổi được chia thành các khoảng 5 năm để tính tỷ lệ hộ nghèo trong từng nhóm tuổi.

Một số nhóm tuổi trẻ có tỷ lệ thuộc hộ nghèo tương đối cao, gồm:

- Nhóm 5–9 tuổi.
- Nhóm 10–14 tuổi.
- Nhóm 15–19 tuổi.

Một số nhóm tuổi rất cao cũng có tỷ lệ nghèo lớn, nhưng quy mô mẫu nhỏ nên cần thận trọng khi diễn giải.

Kết quả này cho thấy mối liên hệ giữa tuổi và tình trạng nghèo có thể không hoàn toàn tuyến tính. Tỷ lệ nghèo thay đổi theo từng giai đoạn vòng đời, nhưng cần kiểm định thêm trước khi khẳng định đây là quan hệ phi tuyến ổn định.

### 2.3. Khám phá mối liên hệ giữa tuổi chủ hộ và quy mô hộ

Dữ liệu cấp hộ được sử dụng để so sánh:

- Nhóm tuổi chủ hộ.
- Số nhân khẩu trong hộ.
- Số trẻ em.
- Số người phụ thuộc.
- Tỷ lệ lao động.
- Phân loại hộ.

Biểu đồ hộp được sử dụng để quan sát sự khác biệt về quy mô hộ giữa các nhóm tuổi chủ hộ và giữa hộ nghèo với hộ cận nghèo.

Phân tích chỉ cho phép kết luận rằng một số nhóm hộ có xu hướng đồng thời xuất hiện chủ hộ trẻ, quy mô nhân khẩu lớn và tỷ lệ phụ thuộc cao. Không thể từ dữ liệu cắt ngang khẳng định rằng chủ hộ trẻ hoặc đông con trực tiếp gây ra tình trạng nghèo.

### 2.4. Khám phá theo huyện và dân tộc

Các đặc điểm tuổi và phân loại hộ được phân tích đồng thời theo huyện và dân tộc nhằm xác định:

- Huyện nào có tỷ lệ hộ nghèo cao.
- Dân tộc nào có tỷ lệ hộ nghèo cao.
- Cơ cấu tuổi của cùng một nhóm dân tộc có khác nhau giữa các địa bàn hay không.
- Những địa bàn có kết quả khác biệt so với xu hướng chung.

Một số địa bàn có thể được xem là trường hợp khác biệt tích cực hoặc tiêu cực. Tuy nhiên, để giải thích tại sao một địa bàn có kết quả tốt hơn địa bàn khác cần bổ sung dữ liệu về thu nhập, sinh kế, giáo dục, hạ tầng và mức độ tiếp cận chính sách.

## 3. Phân tích Chẩn đoán và Kiểm định – Diagnostic Analysis

### Mục tiêu

Tầng chẩn đoán được sử dụng để định lượng mức độ liên hệ giữa các đặc điểm của hộ với trạng thái nghèo, đồng thời đánh giá xem nhóm biến nào có khả năng giải thích sự phân hóa tốt hơn.

Do dữ liệu không chứa đầy đủ thông tin về thu nhập, việc làm, bệnh tật, tài sản và mức độ tiếp cận chính sách, phân tích này chỉ xác định các **yếu tố liên quan**, chưa xác định được nguyên nhân gốc rễ theo nghĩa nhân quả.

### 3.1. Kiểm định mối liên hệ bằng Cramér’s V

Cramér’s V được sử dụng để đo mức độ liên hệ giữa các biến phân loại và `PHAN_LOAI_HO`.

Các biến được kiểm tra bao gồm:

- `NHOM_TUOI_CHU_HO`
- `GIOI_TINH_CHU_HO`
- `DAN_TOC_CHU_HO`
- `HUYEN`
- Các biến phân loại khác ở cấp hộ

Kết quả đối với hai đặc điểm chủ hộ cho thấy:

- `NHOM_TUOI_CHU_HO × PHAN_LOAI_HO`: Cramér’s V khoảng **0,0577**.
- `GIOI_TINH_CHU_HO × PHAN_LOAI_HO`: Cramér’s V xấp xỉ **0,0000**.

Điều này cho thấy nhóm tuổi chủ hộ chỉ có mối liên hệ yếu với trạng thái nghèo, còn giới tính chủ hộ gần như không tạo ra sự phân hóa đáng kể trong tập dữ liệu.

Vì vậy, chưa có cơ sở để khẳng định tuổi chủ hộ là yếu tố quyết định việc một hộ thuộc nhóm nghèo hay cận nghèo.

### 3.2. Đánh giá sức giải thích bằng Logistic Regression

Mô hình Logistic Regression được sử dụng với biến mục tiêu:

\[
Y =
\begin{cases}
1, & \text{Hộ nghèo}\\
0, & \text{Hộ cận nghèo}
\end{cases}
\]

Các nhóm mô hình được xây dựng để so sánh mức độ giải thích:

- Mô hình chỉ sử dụng `HUYEN`.
- Mô hình chỉ sử dụng `DAN_TOC_CHU_HO`.
- Mô hình chỉ sử dụng `NHOM_TUOI_CHU_HO`.
- Mô hình chỉ sử dụng `GIOI_TINH_CHU_HO`.
- Model A: Kết hợp các biến phân loại.
- Model B: Kết hợp biến phân loại và các biến số cấp hộ.

Mức độ phù hợp được đánh giá bằng McFadden Pseudo \(R^2\).

Kết quả trên khoảng **19.504 hộ** cho thấy:

- `HUYEN`: Pseudo \(R^2\) khoảng **0,0886**.
- `DAN_TOC_CHU_HO`: khoảng **0,0628**.
- `NHOM_TUOI_CHU_HO`: khoảng **0,0023**.
- `GIOI_TINH_CHU_HO`: gần **0,0000**.
- Model A: khoảng **0,1210**.
- Model B: khoảng **0,1205**.

Kết quả cho thấy:

1. Yếu tố địa bàn và dân tộc có sức phân hóa lớn hơn tuổi và giới tính chủ hộ.
2. Riêng nhóm tuổi chủ hộ giải thích rất ít sự khác biệt giữa hộ nghèo và hộ cận nghèo.
3. Việc thêm các biến số về cấu trúc nhân khẩu trong Model B chưa cải thiện sức giải thích so với Model A.
4. Các đặc điểm hiện có chỉ giải thích được một phần nhỏ sự phân loại hộ.

Đây là tín hiệu cho thấy trạng thái nghèo còn phụ thuộc vào nhiều yếu tố chưa xuất hiện trong dữ liệu, chẳng hạn như:

- Thu nhập và nguồn sinh kế.
- Trình độ học vấn.
- Việc làm.
- Tình trạng sức khỏe.
- Tài sản và đất sản xuất.
- Khả năng tiếp cận tín dụng.
- Khoảng cách đến trường học, cơ sở y tế và thị trường.
- Mức độ tiếp cận các chương trình hỗ trợ.

### 3.3. Kiểm tra tương quan và đa cộng tuyến giữa các đặc trưng

Các biến số cấp hộ được kiểm tra bằng ma trận tương quan để phát hiện những cặp biến có quan hệ gần tuyến tính.

Một số nhóm biến có khả năng trùng lặp thông tin gồm:

- Số lượng từng nhóm tuổi và tổng số nhân khẩu.
- Tỷ lệ từng nhóm tuổi và số lượng tương ứng.
- `TUOI_LON_NHAT`, `TUOI_NHO_NHAT` và `DO_LECH_TUOI`.
- `SO_PHU_THUOC` và `TY_LE_PHU_THUOC`.
- `SO_LAO_DONG` và `TY_LE_LAO_DONG`.

Các biến có tương quan quá cao cần được xem xét loại bỏ trước khi xây dựng mô hình dự báo nhằm:

- Giảm đa cộng tuyến.
- Tránh lặp lại cùng một thông tin.
- Giúp hệ số mô hình ổn định hơn.
- Hạn chế tình trạng mô hình phức tạp nhưng không cải thiện hiệu quả.

### 3.4. Chẩn đoán chất lượng dữ liệu

Bên cạnh phân tích đặc điểm hộ, nghiên cứu thực hiện kiểm toán chất lượng dữ liệu để đánh giá độ tin cậy của kết quả.

Các nội dung kiểm tra gồm:

- Thiếu `NGAY_SINH` theo huyện.
- Lỗi ngày sinh và tuổi bất thường.
- Trùng lặp nhân khẩu theo mã hộ, họ tên và ngày sinh.
- Trùng lặp thông tin hộ.
- Chất lượng đối chiếu thông tin định danh.
- Số lượng lỗi dữ liệu theo huyện.

Kết quả được tổng hợp trong tệp `quality_review.xlsx`, gồm ba sheet:

1. `Data_Quality`
2. `Missing_DOB_By_Huyen`
3. `Error_By_Huyen`

Phân tích chất lượng dữ liệu giúp xác định những địa bàn có tỷ lệ thiếu hoặc lỗi cao, từ đó hỗ trợ cải thiện quy trình thu thập và rà soát dữ liệu trong các kỳ tiếp theo.

## 4. Phân tích Dự báo – Predictive Analysis

### Mục tiêu

Tầng dự báo là bước phát triển tiếp theo sau khi hoàn thành phân tích mô tả, khám phá và chẩn đoán.

Mục tiêu gồm:

- Phân nhóm các hộ có cấu trúc nhân khẩu tương đồng.
- Dự báo khả năng một hộ thuộc nhóm nghèo thay vì cận nghèo.
- Xác định các đặc điểm đóng góp nhiều nhất vào kết quả dự báo.
- Xây dựng cơ sở cho hệ thống hỗ trợ ra quyết định chính sách.

Các kết quả dự báo chưa được sử dụng để khẳng định hộ sẽ tái nghèo hoặc thoát nghèo theo thời gian, bởi dữ liệu hiện tại chỉ là dữ liệu cắt ngang tại một thời điểm.

### 4.1. Phân cụm hộ gia đình bằng K-Means

K-Means được đề xuất để phân nhóm các hộ dựa trên cấu trúc nhân khẩu, không sử dụng trước nhãn hộ nghèo hoặc hộ cận nghèo để hình thành cụm.

Các biến đầu vào phù hợp gồm:

- `NHAN_KHAU`
- `SO_MAM_NON`
- `SO_HOC_DUONG`
- `SO_LAO_DONG`
- `SO_U50`
- `SO_PHU_THUOC`
- `TY_LE_LAO_DONG`
- `TY_LE_PHU_THUOC`
- `TY_LE_U50`
- `TUOI_CHU_HO`
- `TUOI_TRUNG_BINH_HO`
- `DO_LECH_TUOI`

Các biến phân loại như huyện và dân tộc không nên đưa trực tiếp vào K-Means dưới dạng mã số thứ tự. Nếu cần sử dụng, chúng phải được mã hóa phù hợp hoặc dùng để mô tả các cụm sau khi mô hình hoàn thành.

Quy trình dự kiến:

1. Lựa chọn các biến không bị trùng lặp thông tin.
2. Xử lý giá trị thiếu.
3. Chuẩn hóa dữ liệu.
4. Dùng Elbow và Silhouette Score để lựa chọn số cụm.
5. Xây dựng K-Means.
6. Mô tả đặc điểm của từng cụm.
7. Sử dụng PCA để trực quan hóa cụm trong không gian hai chiều.

Các cụm không được đặt tên trước. Tên gọi như “hộ già hóa”, “hộ gánh nặng học đường” hoặc “hộ có tiềm năng sinh kế” chỉ được sử dụng sau khi quan sát tâm cụm và đặc điểm thực tế.

### 4.2. Phân lớp hộ nghèo và hộ cận nghèo

Bài toán phân lớp được xây dựng với:

- Nhãn 0: `Hộ cận nghèo`
- Nhãn 1: `Hộ nghèo`

Các mô hình dự kiến gồm:

- Logistic Regression làm mô hình cơ sở.
- Decision Tree.
- Random Forest hoặc XGBoost.

Dữ liệu được chia thành tập huấn luyện và tập kiểm tra. Quá trình đánh giá không chỉ sử dụng Accuracy mà còn bao gồm:

- Precision.
- Recall.
- F1-score.
- ROC-AUC.
- Confusion Matrix.

Trong bối cảnh hỗ trợ giảm nghèo, Recall đối với lớp hộ nghèo cần được chú ý vì mô hình bỏ sót một hộ có nguy cơ nghèo có thể làm giảm hiệu quả can thiệp.

Các công cụ như Feature Importance hoặc SHAP có thể được sử dụng để giải thích:

- Đặc điểm nào làm xác suất thuộc hộ nghèo tăng.
- Đặc điểm nào làm xác suất thuộc hộ nghèo giảm.
- Vai trò tương đối của địa bàn, dân tộc và cấu trúc nhân khẩu.

### 4.3. Giới hạn của mô hình dự báo

Mô hình hiện tại chỉ có thể dự báo sự khác biệt giữa hai trạng thái quan sát tại cùng một thời điểm:

- Hộ nghèo.
- Hộ cận nghèo.

Mô hình chưa thể dự báo chính xác việc một hộ sẽ “rơi xuống nghèo”, “tái nghèo” hoặc “thoát nghèo” trong tương lai vì chưa có dữ liệu theo dõi hộ qua nhiều năm.

Để xây dựng hệ thống cảnh báo sớm đúng nghĩa, cần bổ sung:

- Mã hộ ổn định qua các kỳ rà soát.
- Trạng thái nghèo của từng hộ theo từng năm.
- Thu nhập và thay đổi thu nhập.
- Biến động thành viên.
- Việc làm và sinh kế.
- Sức khỏe và chi phí y tế.
- Việc tiếp nhận các chính sách hỗ trợ.

## 5. Ý nghĩa của chiến lược phân tích

Chiến lược nghiên cứu giúp chuyển đổi dữ liệu rà soát hành chính thành bằng chứng phục vụ quản lý thông qua ba giá trị chính.

### Thứ nhất, nhận diện đúng đặc điểm của nhóm hộ nghèo

Kết quả cho thấy cơ cấu tuổi có liên hệ với tình trạng nghèo, nhưng tuổi chủ hộ không phải là yếu tố có sức giải thích mạnh. Sự khác biệt theo địa bàn và dân tộc nổi bật hơn.

### Thứ hai, tránh kết luận nhân quả quá sớm

Dữ liệu chỉ phản ánh mối liên hệ tại một thời điểm. Vì vậy, các kết luận được diễn giải theo hướng:

- “Có xu hướng”.
- “Có mối liên hệ”.
- “Dữ liệu gợi ý”.
- “Có sự khác biệt quan sát được”.

Nghiên cứu không khẳng định rằng tuổi trẻ, đông con, dân tộc hoặc địa bàn trực tiếp gây ra nghèo.

### Thứ ba, định hướng hoàn thiện hệ thống dữ liệu giảm nghèo

Việc mô hình hiện tại chỉ giải thích được một phần trạng thái nghèo cho thấy cần bổ sung dữ liệu kinh tế – xã hội và dữ liệu theo thời gian.

Khi dữ liệu được hoàn thiện, hệ thống có thể hỗ trợ:

- Xác định nhóm hộ cần ưu tiên.
- Phân nhóm chính sách theo đặc điểm hộ.
- Theo dõi sự thay đổi trạng thái nghèo.
- Đánh giá hiệu quả các chương trình hỗ trợ.
- Cảnh báo nguy cơ tái nghèo hoặc nghèo phát sinh.

## 6. Kết luận định hướng

Nghiên cứu triển khai một quy trình phân tích từ dữ liệu nhân khẩu thô đến dữ liệu đặc trưng cấp hộ, sau đó lần lượt thực hiện phân tích mô tả, khám phá và chẩn đoán.

Kết quả bước đầu cho thấy:

- Nhóm hộ nghèo có cơ cấu tuổi trẻ hơn và tỷ lệ trẻ em cao hơn nhóm hộ cận nghèo.
- Nhóm tuổi chủ hộ chỉ có mối liên hệ yếu với phân loại hộ.
- Giới tính chủ hộ gần như không tạo ra sự phân hóa.
- Địa bàn và dân tộc có sức giải thích cao hơn các đặc điểm tuổi riêng lẻ.
- Việc bổ sung các biến số cấu trúc hộ chưa làm tăng rõ rệt sức giải thích của mô hình Logistic Regression.
- Dữ liệu hiện tại chưa đủ để xác định nguyên nhân gốc rễ hoặc dự báo quá trình tái nghèo theo thời gian.
- Chất lượng dữ liệu, đặc biệt là ngày sinh, lỗi dữ liệu và trùng lặp, cần tiếp tục được kiểm soát.

Tầng phân tích dự báo bằng K-Means và các mô hình phân lớp là bước tiếp theo nhằm phân khúc hộ và đánh giá khả năng phân loại. Tuy nhiên, kết quả mô hình cần được xem như công cụ hỗ trợ ra quyết định, không thay thế việc thẩm định thực tế và không được sử dụng như bằng chứng nhân quả.
