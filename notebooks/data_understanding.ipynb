```python
import os
import pandas as pd
from IPython.display import display
from pathlib import Path
ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

folder_path = str(ROOT / "data" / "raw")
files = [
    "HUYỆN CƯ JUT.xlsx", "HUYỆN ĐẮK GLONG.xlsx", "HUYỆN DAK MIL.xlsx",
    "HUYỆN ĐẮK RLẤP.xlsx", "HUYỆN ĐẮK SONG.xlsx", "HUYỆN KRÔNG NÔ.xlsx",
    "HUYỆN TUY ĐỨC.xlsx", "TP GIA NGHĨA.xlsx"
]

print("--- KHẢO SÁT CHI TIẾT TỪNG FILE DATA ---")
print("=" * 60)

for file in files:
    full_path = os.path.join(folder_path, file)
    if os.path.exists(full_path):
        # Đọc dữ liệu từ file huyện
        df = pd.read_excel(full_path)
        huyen = file.replace(".xlsx", "")
        
        print(f"\n★ ĐƠN VỊ: {huyen.upper()}")
        print("-" * 40)
        
        # 1. Kích thước dữ liệu
        print(f"+ Kích thước dữ liệu: {df.shape}")

        # 2. Datatype
        print(f"\n+ Loại dữ liệu:\n{df.dtypes}")

        # 3. Dữ liệu mẫu
        print(f"\n+ Dữ liệu mẫu:")
        display(df.head(2))

        # 4. Thông tin dữ liệu
        print(f"\n+ Info:")
        df.info()
        
        # 5. Tính toàn vẹn dữ liệu
        has_null = df.isnull().sum().any()
        has_nan  = df.isna().sum().any()
        n_duplicated = df.duplicated().sum()
        
        print(f'\nTính toàn vẹn dữ liệu:')
        print(f'+ Có giá trị Null: {has_null}')
        if has_null:
            display(df[df.isnull().any(axis=1)]) 
            
        print(f'+ Có giá trị Nan: {has_nan}')
        if has_nan:
            display(df[df.isna().any(axis=1)])
            
        print(f'+ Số dòng trùng: {n_duplicated}')
        if n_duplicated > 0:
            display(df[df.duplicated()])
            
        print("=" * 60)
```

### Danh sách các vấn đề dữ liệu cần chuẩn hóa

| STT | Vấn đề phát hiện | Chi tiết biểu hiện |
| :--- | :--- | :--- |
| **1** | Tên sheet không nhất quán | Có khoảng trắng thừa, cách viết tắt khác nhau giữa các huyện. |
| **2** | Dòng header lệch vị trí | Dòng header không nằm ở dòng đầu tiên của file. |
| **3** | Cột tên không chuẩn | Chứa ký tự xuống dòng (`\n`), khoảng trắng thừa, viết chữ thường. |
| **4** | Ngày sinh sai định dạng | Lưu nhiều định dạng khác nhau: `DD/MM/YYYY`, `YYYY-MM-DD`, chỉ có năm, hoặc dạng số timestamp. |
| **5** | Số CCCD/CMND bị biến dạng | Bị lưu dưới dạng số float (`.0`), ký hiệu khoa học (`6.73e+10`), hoặc chứa text rác. |
| **6** | Tên dân tộc không đồng nhất | Viết sai chính tả, viết tắt hoặc nhập mã số thay vì tên dân tộc. |
| **7** | Biến thể chuỗi ở kết quả tra cứu | Có nhiều biến thể chuỗi khác nhau cho cùng một kết quả, cần phải gom nhóm lại. |
